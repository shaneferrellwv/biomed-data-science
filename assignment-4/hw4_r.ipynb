```r
annual <- read.csv("county_annual.csv", stringsAsFactors = FALSE)
land <- read.csv("land_area.csv", stringsAsFactors = FALSE)
population <- read.csv("county_pop_arcos.csv", stringsAsFactors = FALSE)
```

```r
annual$countyfips[annual$BUYER_STATE == "AR" & annual$BUYER_COUNTY == "MONTGOMERY"] <- "05097"
annual <- annual[!is.na(annual$BUYER_COUNTY), ]
```

```r
land_area <- land[, c("Areaname", "STCOU", "LND110210D")]
names(land_area)[names(land_area) == "STCOU"] <- "countyfips"
annual$countyfips <- sprintf("%05d", as.integer(annual$countyfips))
population$countyfips <- sprintf("%05d", as.integer(population$countyfips))
land_area$countyfips <- sprintf("%05d", as.integer(land_area$countyfips))
county_info <- merge(population, land_area, by = "countyfips", all.x = TRUE)
opioid <- merge(annual, county_info, by = c("countyfips", "year"), all.x = TRUE, suffixes = c("_annual", "_population"))
```
