## 1) Sorting Hat Program

```python
def sorting_hat():

    gryffindor_score = 0
    slytherine_score = 0

    answer = input("Is a hot dog a sandwich? (Y/N): ")
    if answer == "Y":
        gryffindor_score = gryffindor_score + 1
    elif answer == "N":
        slytherine_score = slytherine_score + 1

    answer = input("Does pineapple go on pizza? (Y/N): ")
    if answer == "Y":
        gryffindor_score = gryffindor_score + 1
    elif answer == "N":
        slytherine_score = slytherine_score + 1

    answer = input("Would you choose to fight one horse-sized duck over 100 duck-sized horses? (Y/N): ")
    if answer == "Y":
        gryffindor_score = gryffindor_score + 1
    elif answer == "N":
        slytherine_score = slytherine_score + 1
    
    return "Gryffindor" if gryffindor_score >= slytherine_score else "Slytherine"
```

## 2) mydata Class

```python
class mydata:

    def __init__(self, numbers):
        self.numbers = numbers

    def mean(self):
        return sum(self.numbers) / len(self.numbers)

    def std(self):
        return (sum((x - self.mean()) ** 2 for x in self.numbers) / (len(self.numbers))) ** 0.5
```

```python
x = mydata([1, 2, 7, 5])
print(x.mean())
print(x.std())
```

## 3) lag Function

```python
import numpy as np
```

```python
def lag(a, n):
    return [np.nan] * n + a[0:-n]
```

```python
a = [1, 6, 8, 4]
print(lag(a, 1))
print(lag(a, 2))
```
